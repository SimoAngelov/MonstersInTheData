# Project Proposal
*Version 1.0 – 25 November 2026*

## Title
**Spatial and Temporal Patterns in Reported Cryptid Sightings: A Comparative Statistical and Geospatial Analysis of Bigfoot and Loch Ness Reports**

## 1. Problem Statement
Reported sightings of cryptids such as Bigfoot (Sasquatch) and the Loch Ness Monster represent large collections of anecdotal, crowdsourced observations. These reports exhibit clear spatial and temporal structure, yet the extent to which this structure can be explained by geographic accessibility, human population density, environmental features, and reporting behaviour remains insufficiently quantified using rigorous statistical and geospatial methods.

This project treats the reports strictly as data on human reporting patterns, not as evidence for the existence of unknown animals. Following the framing of Paxton et al. (2025), the data support inferences about the *population of reports*, not about the animals themselves.

The project follows the scientific method: question → research → hypotheses → data → exploration → testing → conclusions → communication.

## 2. Research Question
**Can the observed spatial, temporal and environmental patterns in reported Bigfoot and Loch Ness sightings be largely explained by geographic accessibility and human reporting factors?**

## 3. Objectives
1. Collect, clean and harmonise two independent cryptid sighting datasets (BFRO Bigfoot reports and the Loch Ness Monster reports database) together with a control dataset of real biological observations (GBIF) under a common tidy schema.
2. Quantify spatial clustering, seasonal patterns and temporal trends in the cryptid reports.
3. Test associations between sighting density and geographic/human factors (population density, forest cover, road access), accounting for observer effort.
4. Compare the spatial and temporal signatures of cryptid reports with those of selected real species from GBIF in the same regions.
5. Clearly document data limitations, biases and the boundaries of inference.
6. *(Optional)* Extract structured information from the free-text report descriptions and test whether report content differs by quality class and season.

All sources will be cited and their known biases documented (self-selection, investigation filters, media influence, varying data quality over time).

### Integration strategy
- **Tidy common schema** for all sighting/occurrence sources (columns = variables, rows = observations): `source, date, year, month, latitude, longitude, location_precision, quality_class`. Fields that a source does not have (e.g. coordinates for Loch Ness) are left empty, not imputed.
- **Merge (North America):** BFRO + GBIF + covariates aggregated to a common spatial unit (county or regular grid); ID and type compatibility checked after every merge.
- **Temporal comparison (Scotland & North America):** harmonised monthly and yearly series for BFRO, Loch Ness and GBIF.
- The whole pipeline from raw to processed data is automated in code and every transformation is logged, so the analysis is reproducible.

Because Loch Ness is a single ~37 km lake and the database contains no coordinates, it is analysed **temporally**, not with density/clustering methods. The two cryptid datasets are linked through the common schema and through comparable quality measures (BFRO class A/B vs. first-/second-hand Loch Ness reports).

## 5. Hypotheses
α = 0.05 unless stated otherwise. Hypotheses H1–H7, including the H5 event dates, are fixed **before** the data are analysed (confirmatory); any additional pattern found during the analysis is reported as exploratory.

| ID | Alternative hypothesis (H₁) | Null hypothesis (H₀) |
|----|-----------------------------|----------------------|
| H1 | Bigfoot reports are more spatially concentrated than expected from human population and road access alone. | BFRO reports are distributed as expected under the population/access null model (sections 6.3–6.4). |
| H2a | Report counts vary by month. | Report counts are uniformly distributed across months (after correcting for month length). |
| H2b | Within the same region and overlapping years, the monthly profile of cryptid reports is positively correlated with the monthly profile of GBIF observations (i.e. both track observer effort). | The monthly profiles are not positively correlated (ρ ≤ 0). |
| H3 | Bigfoot report density per county is positively associated with forest cover and road access after accounting for population. | There is no positive association between BFRO report density and forest cover/road access after accounting for population. |
| H4 | Bigfoot reports overlap spatially with American black bear occurrences (GBIF) more than expected by chance. | The overlap is not greater than expected under the spatial null model (section 6.4). |
| H5 | Loch Ness report frequency increases after pre-specified events: 1933–34 (new A82 road along the north shore, first press coverage, the "surgeon's photograph"), treated as one event because the years are adjacent; 1962 (start of organised watches by the Loch Ness Phenomena Investigation Bureau). | Report frequency is the same before and after each pre-specified event. |
| H6 | Report characteristics differ by report quality **within each dataset**: BFRO class A vs B (distance to nearest road, season); Loch Ness first- vs second-hand reports (month, observer location, reported distance to the object). | The distributions of the selected characteristics do not differ between quality categories. |
| H7 *(optional, Bigfoot only)* | The content of BFRO report descriptions (witness activity – hunting, camping, driving; mentions of bears, sounds, smells) differs by class and season. | There is no association between the extracted text features and class or season. |

Notes:
- **H2 – comparable periods:** most Loch Ness reports pre-date 2000 (median year 1963), while GBIF records are mostly recent. Seasonal comparisons are therefore made only within overlapping periods and the same region; where no overlap exists, each dataset's seasonality is tested on its own. Day length is analysed as a separate explanatory variable, not as a substitute for season. A non-significant H2b test is **not** evidence that the profiles are similar; similarity is claimed only from a positive correlation (H2b) together with its confidence interval.
- **H5 – which year:** the year of the encounter is used as primary; the year of the report is used as a sensitivity check, because retrospective reports collected after an event (e.g. by the Bureau after 1962) can create artificial jumps.
- **H6 – different distances:** BFRO "distance to road" (computed) and Loch Ness "distance to object" (reported by the witness) are different quantities and are never compared with each other.

## 6. Methodology
The analysis has a **core** built on the methods from the course lectures, and an **extension** with specialised spatial and count-data methods. The core alone answers every hypothesis; the extension refines the answers if time allows.

### 6.1 Data pipeline
1. **Collection & consolidation** – read each source, check `shape`, `head()` and `dtypes`; record provenance.
2. **Tidying** – reshape to the common schema (melt/pivot/merge/split as required).
3. **Cleaning & quality assessment** – duplicates, missing values, coordinate precision, date validity; completeness by period and region. Replacing or correcting values is done sparingly and every change is logged.
4. **Outliers** – inspected individually; key analyses are run **with and without** outliers and compared.

### 6.2 Exploratory Data Analysis (Tukey)
Univariate distributions first (histograms, boxplots), then breakdown by source, region, class and period, then grouping and correlations. EDA is used to refine the hypotheses and check test assumptions, without losing sight of the original research question.

Visualisation rules: appropriate chart type, labelled axes, titles and legends, no pie charts, histograms used only for distributions, colour and marker size used to carry information.

### 6.3 Core methods (from the lectures)

| Hypothesis | Core method |
|------------|-------------|
| H1 | χ² goodness of fit: observed reports per region vs. expected counts $E_i = N \cdot w_i / \sum_j w_j$, where the weight $w_i$ is population (baseline) or population × road density (access model). Regions are aggregated so that every $E_i \geq 5$. χ² detects any departure; the direction (over-concentration) is shown by standardised residuals and the variance-to-mean ratio |
| H2a | χ² goodness of fit for months (with month-length correction), separately for each dataset |
| H2b | Spearman correlation between the 12 monthly shares of cryptid reports and of GBIF observations (same region, overlapping years), one-sided, with a confidence interval |
| H3 | Report rate per 100,000 inhabitants per county; Spearman and partial Spearman correlation with forest cover and road density (controlling for population density); scatterplots. Counties with very small populations are flagged, because their rates are unstable |
| H4 | Spearman correlation of gridded Bigfoot and black bear densities, reported as a descriptive measure; its ordinary p-value is too optimistic because of spatial autocorrelation (valid test in 6.4) |
| H5 | Confirmatory: Poisson rate test comparing yearly counts in fixed windows (e.g. 10 years) before and after each pre-specified event, with a confidence interval for the rate ratio; moving averages |
| H6 | Mann–Whitney U (distances are skewed; t-test only if assumptions hold): distance to roads for class A vs B, reported distance for first- vs second-hand; χ² test of independence (quality × season, quality × observer location) |

For every result: test statistic, p-value, effect size **and confidence interval** (e.g. for proportions and mean differences). Results are interpreted carefully: p-value is the probability of data at least this extreme if H₀ is true; failing to reject H₀ does not prove H₀ ("absence of evidence is not evidence of absence"); correlation does not imply causation. Multiple tests are adjusted with the Holm method. Core spatial tests (H1, H3, H4) assume independent observations; because neighbouring areas are correlated, their p-values are treated as optimistic and checked with the extension methods where possible.

### 6.4 Extension methods

| Hypothesis | Extension method | Why |
|------------|------------------|-----|
| H1 | Kernel density estimation; nearest-neighbour index and inhomogeneous Ripley's K; null model simulated by placing the same number of points at random with intensity ∝ population × road density (Monte Carlo envelopes) | Tests clustering at multiple distances, not just per region |
| H2a | Rayleigh test (circular statistics), on day of year where the exact date is known, otherwise on months with a grouping correction | Treats December and January as neighbours |
| H3 | Poisson → negative binomial regression with log(population) offset; Moran's I on residuals | Correct model for overdispersed counts; checks spatial dependence |
| H4 | Main test – spatially restricted permutation test: the bear density grid is randomly shifted (toroidal shift) within the study area, preserving its spatial autocorrelation, to build the null distribution of the overlap statistic | Accounts for spatial autocorrelation |
| H5 | Change-point / segmented trend analysis against a constant-rate Poisson baseline – **exploratory only** | Finds breaks without fixing dates in advance; reported separately from the confirmatory test |

Each extension method is explained with its formulas, assumptions and references, and compared to the corresponding core method.

### 6.5 Text as data (optional, H7)
Bag-of-words, n-grams and TF-IDF are statistical representations of text, not machine-learning models; no language models or text generation are used.

| Step | Method |
|------|--------|
| Information extraction | Regular expressions / keyword lists on BFRO descriptions → new categorical columns (time of day, witness activity, mentions of bears, sounds, smells) |
| Testing extracted features | χ² test of independence (feature × class, feature × season), with confidence intervals for proportions |
| Distinctive vocabulary | Bag-of-words with unigrams and bigrams (e.g. "wood knock"); TF-IDF, $\text{tf-idf}(t,d) = \text{tf}(t,d)\cdot\log\frac{N}{\text{df}(t)}$, comparing class A vs B and summer vs winter |

The Loch Ness database contains no free-text descriptions, so this part covers Bigfoot reports only. If time does not allow, this section moves to Future Work.

## 8. Limitations
- All cryptid data are anecdotal and subject to strong reporting and selection biases.
- GBIF data are also observer-biased; they serve as a comparison of reporting processes, not as ground truth.
- Data quality and completeness vary substantially across time and regions; the Loch Ness data have no coordinates and many missing values (e.g. time of day 39% complete).
- The public Loch Ness file is a reduced version (1,433 of 1,800+ reports); the BFRO export is a snapshot and covers only the publicly available reports.
- No causal claims about the existence of cryptids will be made.
- Results concern patterns in *reported* observations only.
- Spatial units are not independent; tests without a spatial null model give optimistic p-values.
- Bigfoot vs black bear overlap would be consistent with misidentification but cannot prove it.
- Report texts are analysed but not republished in full; only short excerpts are quoted, subject to the dataset licence.

## 9. Future Work
The framework can be extended to other sighting datasets once sufficiently structured and geocoded data are available. The strongest candidate is the Tasmanian Thylacine Sighting Record Database (Brook et al., 2023/2024; 1,223 quality-rated, geo-located records), which would allow the full spatial pipeline to be repeated in a second region. Other candidates (Dogman, Yeti, Orang Pendek, eastern cougar) currently lack structured public data.
