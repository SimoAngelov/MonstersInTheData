## Do Cryptid Sightings Follow People or Monsters? A Look at Bigfoot and Loch Ness Reports

MonstersInTheData
*Version 1.2, 25 November 2026. Shortened and rewritten; the hypotheses are unchanged from version 1.1.*

## Why this topic
As a kid I read a lot of mystery books about cryptids, and in seventh grade I watched The Secret Saturdays on Cartoon Network. Since then the unknown has been my favourite subject.

I have never looked at cryptozoology from the research side, so I am curious what the data will show. My guess is that most sightings are animals that people misidentified. What interests me more is why so many people say they have seen a cryptid at all.

## Research question
Can the patterns in reported Bigfoot and Loch Ness Monster sightings (where, when in the year, and in which years) be explained mostly by where people are, when they spend time outdoors and what the media talks about?

I treat the reports as data about the people who report them, not as evidence that the animals exist. Paxton et al. (2025) say it well: this kind of data tells us about the population of reports, not about the monster.

## Data
| Source | What it contains | Role |
|--------|------------------|------|
| BFRO Bigfoot reports (TidyTuesday export by T. Renner) | About 5,000 reports with date, county, coordinates (for part of them), class A/B/C and the full report text | Main dataset, space and time |
| Loch Ness Monster reports (Paxton, Shine & Popov, 2025; GPL-3) | 1,433 reports with year, month, where the observer stood, distance, and first- or second-hand report. No coordinates, no text | Second dataset, time only |
| GBIF occurrences | American black bear in North America, Eurasian otter around Loch Ness | Control: real animals, also reported by people |
| US Census, NLCD, TIGER/OpenStreetMap | County population, forest cover, roads | Explanatory variables |

Loch Ness is one lake about 37 km long and the reports have no coordinates, so I only analyse it over time. The two cryptid datasets are linked through a common table format (source, date, year, month, coordinates, quality) and through their quality measures: BFRO class A vs B and first- vs second-hand Loch Ness reports.

All raw data are tracked with DVC and imported from their original public sources.

## Hypotheses
I fixed these before analysing the data. To be transparent: before writing them I looked at the column list and the share of missing values in the Loch Ness file, but at no results. H5 is a replication rather than a new idea, because Hotham (2026) already reported a jump in Loch Ness reports in 1933–34.

| | Hypothesis (H₁) |
|---|---|
| H1 | Bigfoot reports are more concentrated in space than population and road access alone would predict. |
| H2a | The number of reports changes with the month. |
| H2b | In the same region and years, the monthly pattern of cryptid reports is positively correlated with the monthly pattern of GBIF observations, i.e. both follow when people are outdoors. |
| H3 | Counties with more forest and more road access have more Bigfoot reports per person. |
| H4 | Bigfoot reports overlap with black bear observations more than chance would explain. |
| H5 | Loch Ness reports become more frequent after 1933–34 (new road along the north shore, first newspaper stories, the "surgeon's photograph") and after 1962 (organised watches by the Loch Ness Phenomena Investigation Bureau). |
| H6 | Inside each dataset, reports differ by quality: BFRO class A vs B in distance to the nearest road and season; Loch Ness first- vs second-hand reports in month, observer location and reported distance. |
| H7 (optional) | The words in BFRO reports (what the witness was doing, mentions of bears, sounds or smells) differ by class and season. |

In each case the null hypothesis H₀ is "no difference" or "no association". For H2b it is "the monthly patterns are not positively correlated" (ρ ≤ 0). A non-significant result does not show that the patterns are similar, and I will write "do not reject H₀", never "accept H₀".

Two things I will keep apart: the BFRO distance is the distance to the nearest road, which I calculate, while the Loch Ness distance is what the witness reported. They are different quantities and I will not compare them. For H5 I use the year of the sighting, and the year of the report only as a check, because about 14% of Loch Ness reports were made more than a year later.

## Methods
I will use the methods from the course:

1. **Cleaning.** Bring all sources into one tidy format, check `shape`, `head()` and `dtypes`, and log every step: what I removed, how many rows, and why.
2. **EDA.** One variable at a time first (histograms next to boxplots), then by source, region, class and period.
3. **Tests.**
   - H1: chi-square goodness of fit, observed reports per region against the counts expected from population (and from population × road density).
   - H2a: chi-square goodness of fit for months, corrected for month length. H2b: Spearman correlation of the 12 monthly shares.
   - H3: Spearman and partial Spearman correlation of reports per 100,000 people with forest cover and road density.
   - H4: Spearman correlation of Bigfoot and bear densities on a grid. Neighbouring cells are not independent, so I treat its p-value as too optimistic.
   - H5: comparison of yearly report rates in fixed windows before and after each event.
   - H6: Mann–Whitney U test for distances, chi-square test of independence for categories.
   - H7: keywords extracted with regular expressions and tested with chi-square; TF-IDF for the most typical words per class.

For every test I report the statistic, the p-value, the effect size and a confidence interval. I report all tests, including the ones that come out non-significant, and correct for multiple testing with the Holm method.

If time allows, I will add methods built for spatial and count data (point-pattern tests, negative binomial regression). Everything runs from code, so the analysis can be repeated with `dvc repro`.

## Limitations
- All cryptid reports are anecdotes, and who reports and why is strongly biased.
- GBIF data are also collected by people, so I use them as another reporting process, not as the truth.
- Data quality changes a lot over time and between regions. The Loch Ness data have no coordinates and many missing values.
- The public Loch Ness file is a reduced version (1,433 of more than 1,800 reports).
- Neighbouring regions are not independent, so simple spatial tests are too optimistic.
- An overlap with black bears would fit the idea that people mistake bears for Bigfoot, but it cannot prove it.
- I make no claims about whether cryptids exist.
